### Question A (Ex2, part of Q1 and Q2 of TD1) - Historical VaR (Non-parametric)

The objective of this question is to estimate the market risk of the Natixis stock using a historical, non-parametric approach(VaR).

<span style="color:#7a3e0b"> All VaR estimates in this section use a tail probability (risk level) α = 5%, i.e. a 95% confidence level.</span>

#### I. Data preparation and return computation

We start by loading the Natixis price data (between January 2015 and December 2016) provided for the project and computing daily returns from prices just after transforming the date and chronological ordering.

<span style="color:#7a3e0b">The same dataset "Natixis_stock" and the same return series will be used in Questions B and C. Therefore, we prepare thedata once, and no further changes are needed.</span>

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

#data = pd.read_csv("../data/Natixis_stock.txt", delimiter='\t', 
                   #header=None, parse_dates=[0]) #we will mainly use main dataframe to analyze data in our whole project
data = pd.read_csv("../data/Natixis_stock.txt", delimiter='\t', 
                   header=None, parse_dates=[0])
data.columns = ['Date', 'Value'] # rename columns
data['Date'] = pd.to_datetime(data['Date'], format="%d/%m/%Y") 

data['Value'] = data['Value'].str.replace(',', '.').astype(float) #retype float number as we read a txt
data = data.sort_values('Date') # let's sort by date to calculate returns 

def Returns(data): #function to go to the next row in the same column(we didn't use pct_change() a pandas functiob)
    data['Returns'] = (data['Value'] - data['Value'].shift(1)) / data['Value'].shift(1)
    return data

data = Returns(data)
print(data)

#the first value of returns is NaN which is logical so we decided to erase the row from our dataset
data = data.dropna()
print(data)


alpha = 0.05 # tail probability / risk level (=>95% confidence level)
value_portfolio = 100000 #value portfolio is the actual value of the portfolio
```

Bonus: we represent the evolution of the Natixis stock price over time to better understand the data we are working with.

The asset price moves up and down in a typical way, which is why risk needs to be measured.

#### II. Historical VaR estimation (Question A.a)

We now estimate the historical Value at Risk using a non-parametric approach based on kernel density estimation, as introduced during the lectures.

```python
# on our sorted dataset we select the time periods required to estimate and validate the historical Var.
data_date_sort_2015 = data[(data["Date"] >= '2015-01-01') & (data["Date"] <= '2016-12-31')]
data_date_sort_2017 = data[(data["Date"] >= '2017-01-01') & (data["Date"] <= '2018-12-31')]

#Once the selecton is done now we can sort returns
data_date_sort_2015 = data_date_sort_2015.sort_values("Returns") # using build in sort function of pandas
data_date_sort_2017 = data_date_sort_2017.sort_values("Returns")

print(data_date_sort_2015) #quick check
```

<span style="color:#1f4fd8"> We recall the definition of the historical Value at Risk as introduced during the lectures (see slide on Value at Risk, page 68). </span>

<span style="color:#7a3e0b">We estimate the VaR as the empirical α quantile of the return distribution(definition in the lectures)</span>

```python
#historical VaR estimation using the empirical quantile
var_historical = np.quantile(data_date_sort_2015["Returns"], alpha)

var_historical_value = - var_historical * value_portfolio # we have VaR expressed in monetary value
var_historical, var_historical_value
```

This value corresponds to the 5% quantile of daily returns (tail probability α = 5%), i.e. the one-day VaR at 95% confidence level, based on historical returns observed between 2015 and 2016.

<span style="color:#7a3e0b"> Note on the sign: we work with returns, so the VaR we report is a negative return threshold (for example -3.78%). 
If we want the VaR as a positive loss number (in % or in euros), we just take -VaR. that’s why we use - var_historical when we compute the monetary VaR.</span>

We now refine the estimation by using a kernel based non-parametric approach, as required in the project instructions

We use a biweight kernel to estimate the return distribution. <span style="color:#1f4fd8"> The kernel density estimation is used to approximate the distribution and extract the corresponding quantile, as introduced during the lectures (see slide on VaR estimation, page 69). </span>

<span style="color:#7a3e0b">The biweight kernel used in the following estimation is directly specified in the project statement.</span>


```python
import numpy as np

def kernel(u):  # biweight pdf
    u = float(u)
    if abs(u) <= 1:
        return (15/16) * (1 - u*u)**2
    return 0

def kernel_cdf(u):  # biweight CDF (intégrale exacte du noyau)
    u = float(u)
    if u <= -1:
        return 0
    if u >= 1:
        return 1
                                   # 0,5 + ∫0^u (15/16)(1-t^2)^2 dt
    return 0.5 + (15/16)*u - (5/8)*u**3 + (3/16)*u**5

def bandwidth_rule_of_thumb(returns):
    r = np.asarray(returns, dtype=float)
    n = r.size
    sigma = r.std(ddof=1) 
    return sigma * (4/(3*n))**0.2  # ≈1.06*sigma*n^(-1/5)

def kde_cdf(returns, x, h):  # compute the cumulative probability at x and we average the kernel CDF for each return
    r = np.asarray(returns, dtype=float)
    u = (x - r) / h 
    return np.mean([kernel_cdf(ui) for ui in u])

def VaR_Kernel(data, alpha=0.05): 
    r = np.asarray(data["Returns"], dtype=float)
    h = bandwidth_rule_of_thumb(r) #find the point where the estimated CDF reaches alpha "rule of thumb"
    lo = r.min() - 3*h
    hi = r.max() + 3*h     # search interval for the VaR

    for _ in range(60): # binary search to find the quantile
        mid = 0.5 * (lo + hi)
        if kde_cdf(r, mid, h) < alpha:
            lo = mid
        else:
            hi = mid
    return 0.5 * (lo + hi)

Value_at_Risk = VaR_Kernel(data_date_sort_2015, 0.05)
print(f"Value at Risk of Kernel : {Value_at_Risk}")
```

<span style="color:#7a3e0b"> This kernel VaR is just a return quantile, so it is negative by construction. If we talk in terms of loss size, it corresponds to a one-day VaR of about 3.81% (i.e., -VaR). </span>


Bonus: We display the kernel density estimate together with the return distribution and the estimated VaR .

```python
plt.figure(figsize=(10, 5)) #we only use this plot to better understand the results obtained above

plt.hist(data_date_sort_2015["Returns"], bins=40, density=True, alpha=0.5, label="Returns")

x_values = np.linspace(
    data_date_sort_2015["Returns"].min(),
    data_date_sort_2015["Returns"].max(),
    500
)

kernel_values = [kernel_density(data_date_sort_2015, x) for x in x_values] # kernel density evaluated on the grid

plt.plot(x_values, kernel_values, label="Kernel density")
#vertical line corresponding to the estimated VaR
plt.axvline(Value_at_Risk, linestyle="--", label="Kernel-based VaR")

plt.xlabel("Returns")
plt.ylabel("Density")
plt.title("Kernel-based Value at Risk")
plt.legend()
plt.show()
```

<span style="color:#0b7a3e"> From the graph, we can see that the estimated VaR is in the left tail of the return distribution and remains very close to the empirical VaR computed previously which confirms the consistency of the historical risk estimation. </span>

### Question B (Ex2, Q5 of TD2) - Expected Shortfall

In this question we compute the Expected Shortfall associated with the non-parametric VaR estimated in Question A.

<span style="color:#7a3e0b"> We reuse the same return series as in Question A and the same non-parametric VaR threshold </span>

Since the VaR only gives a quantile of the loss distribution, we also compute the Expected Shortfall to measure the average loss conditional on returns falling below the VaR threshold.  
So we simply average the returns that fall below the VaR threshold.


<span style="color:#7a3e0b">sign convention:</strong> VaR and ES take negative values since risk measures are expressed in terms of returns.</span>

<span style="color:#1f5faa"><strong>Course reference:</strong> The Expected Shortfall (also AVaR or conditional VaR) is introduced in the lectures as the average loss given we are in the VaR tail, i.e. when returns fall below the VaR threshold (see slide “Expected shortfall”, page 75).</span>

<span style="color:#1f5faa"> In the lectures, Expected Shortfall is presented as a spectral (distortion) risk measure. For a continuous distribution, it can be written as:
ES_alpha = E[R | R <= VaR_alpha]. This is the definition we use here.</span>


We now compute the Expected Shortfall associated with the non-parametric VaR estimated in Question A.

#### Computation of the Expected Shortfall :

In practice, we simply average the returns that are below the VaR level estimated in Question A.

<span style="color:#7a3e0b">VaR and ES are reported as returns(negative quantities).</span>

```python
# select the returns that are below (or equal to) the VaR threshold (worst tail)
rdt_over_VaR = data_date_sort_2015["Returns"][data_date_sort_2015["Returns"] <= Value_at_Risk]
#we compute ES empirically:we average the returns in the left tail (R <= VaR_alpha)   
# which matches ES_alpha = E[R|R <= VaR_alpha].
ES = np.mean(rdt_over_VaR)

print("Non parametric VaR:", Value_at_Risk)
print("Expected Shortfall (non-parametric):", round(ES, 5))
```

<span style="color:#0b7a3e">Here, the VaR (95%) is about -3.81% while the ES (95%) is about -5.34%. This makes sense: ES averages the worst returns in the 5% tail, so it tells us what the loss looks like on average when we are beyond the VaR threshold (in loss terms, about 5.00% on those days).</span>


### Question C (Ex2, Q1 and Q2 of TD3) - Extreme Value Theory (EVT)

In this question, we focus on the analysis of extreme returns using tools from Extreme Value Theory (EVT).  
We rely on the Natixis return series and study separately extreme gains and extreme losses in order to better understand tail behavior.

<p><span style="color:#7a3e0b">We use the same return series as in Question A, but now we zoom in on the extreme tails with an EVT approach.</span></p>

<p><span style="color:#7a3e0b"><strong>Sign convention:</strong> in this section we work in loss terms. When returns are negative, we define losses = -returns, so losses are positive numbers. 
This means EVT VaR is a positive loss. If we want to compare with the previous sections (VaR on returns), we just take the negative of it.</span></p>

#### I. Data preparation and tail separation (EVT setup, before Question C.a)

For to apply Extreme Value Theory, we in first split the return distribution into two tails corresponding to extreme gains and extreme losses, which are analyzed separately. Extreme losses are defined as \(L_t = -r_t\) so EVT is applied to the right tail of istribution.

```python
gains = data['Returns'][data['Returns'] > 0]
losses = -data['Returns'][data['Returns'] < 0]#converts losses into positive values thus EVT applies to the right tail.

gains_sorted = np.sort(gains)
losses_sorted = np.sort(losses) 
```

#### II. Tail index estimation with the Pickands estimator (Question C.a)

In this part, we estimate the tail index of the return distribution using Extreme Value Theory. We rely on the Pickands estimator to characterize the behavior of extreme gains and extreme losses separately.

<span style="color:#1f5faa"><strong>Course reference:</strong>
We rely on the Pickands estimator introduced in the Extreme Value Theory lectures to estimate the tail index ξ of the return distribution. The following formula defines the estimator used in our implementation (EVT-based VaR estimation slides, page 180).</span> 

<span style="color:#1f5faa"><strong>Course reference:</strong>
The sign of the tail index ξ allows us to characterize the type of extreme value distribution. ξ > 0 corresponds to a Fréchet distribution (heavy tails), ξ = 0 to a Gumbel distribution, and ξ < 0 to a Weibull distribution. We rely on this classification to interpret our estimates (see EVT lectures, same section, page 166).</span> 

We now apply the Pickands estimator separately to extreme gains and extreme losses.

```python
def pickands_estimator(extreme_values):
 
    #Pickands estimator for the tail index, extreme observations sorted in increasing order.
    n = len(extreme_values)
    
# we set k = int(log(n)) (rule of thumb from the lectures) to choose how many tail points we use.
    k = int(np.log(n))
    x1 = extreme_values[n - k]  #required order statistics
    x2 = extreme_values[n - 2 * k]
    x3 = extreme_values[n - 4 * k]
    
    # we seek the Pickands estimator according to the theoretical formul
    xi_hat = (1 / np.log(2)) * np.log((x1 - x2) / (x2 - x3))
    return xi_hat

# we estimate the tail index separately for gains and for losses
xi_gains = pickands_estimator(gains_sorted)
xi_losses = pickands_estimator(losses_sorted)
xi_gains, xi_losses
```

<span style="color:#0b7a3e">We obtain a positive tail index for extreme gains (ξ ≈ 0.58) and a negative tail index for extreme losses (ξ ≈ −0.51) which indicates asymmetry in the tail behavior of Natixis returns. Extreme positive returns exhibit heavy-tailed behavior (Fréchet type distribution) while extreme losses appear to be lighter-tailed (Weibull type distribution_. In practice, this suggests that very large upward movements may occur with more pronounced extremes than downward movements in terms of tail thickness.</span>

<span style="color:#7a3e0b">As in standard EVT applications we suppose returns are independent and identically distributed. Moreover, since the Pickands estimator uses only a limited number of extreme observations, the tail index estimates can be sensitive to sampling variability and to the choice of the intermediate sequence k.</span>

Bonus: As a visual complement, we plot the empirical distributions of extreme gains and extreme losses. This allows us to illustrate the difference in tail behavior highlighted by the Pickands estimates.

```python
plt.figure(figsize=(8, 5))

plt.hist(gains_sorted, bins=40, density=True, alpha=0.7,
         label="Extreme gains", edgecolor="black")

plt.hist(losses_sorted, bins=40, density=True, alpha=0.7,
         label="Extreme losses", edgecolor="black")

plt.xlabel("Return magnitude")
plt.ylabel("Density")
plt.title("Empirical distribution of extreme gains and losses")

plt.legend()
plt.grid(True)
plt.show()
```

<span style="color:#0b7a3e"><strong>Comment:</strong>
The empirical distributions confirm the asymmetric tail behavior( more pronounced extreme gains than extreme losses)</span>

#### III. EVT  Value at Risk estimation (Question C.b)

In this part, we use Extreme Value Theory to compute Value at Risk estimates based on the tail behavior(characterized before using the Pickands estimator)

We focus on the left tail of the return distribution (losses) since Value at Risk is a downside risk measure. The EVT-based VaR is computed for several confidence levels.

<span style="color:#7a3e0b">The EVT VaR is computed using the tail index estimates obtained in Question C.a. We assume that returns are independent and identically distributed.</span>

<span style="color:#1f5faa"><strong>Course reference:</strong> We use the EVT-based VaR formulation introduced in the lectures where the Value at Risk is expressed using the Pickands estimator of the tail index ξ (see EVT-based VaR estimation slides, page 187).</span>

We now compute the EVT Value at Risk for several confidence levels and compare it to the historical VaR obtained in Question A.

<span style="color:#7a3e0b">Like before we work with extreme losses expressed in positive terms. This makes the EVT Value at Risk easier to interpret as a loss threshold and avoids sign mistakes in the implementation.</span>

```python
def var_evt_pickands(extreme_losses, xi_hat, alpha):
   
    n = len(extreme_losses)
    #we keep the same choice of k to ensure consistency between the two steps
    k = int(np.log(n))
    # tail order statistics required by the EVT formula
    x_k = extreme_losses[n - k]
    x_2k = extreme_losses[n - 2 * k]

    numerator = (k / (n * (1 - alpha))) ** xi_hat - 1
    denominator = 1 - 2 ** (-xi_hat)
    var_evt = (numerator / denominator) * (x_k - x_2k) + x_k
    return var_evt
```

```python
confidence_levels = [0.90, 0.95, 0.99] # the  several confidence levels

for alpha in confidence_levels:
    var_alpha = var_evt_pickands(losses_sorted, xi_losses, alpha)
    print(f"EVT-based VaR (Pickands) at {int(alpha*100)}% confidence level: {var_alpha:.5f}")
```

<p><span style="color:#0b7a3e">Uwe can see the VaR going up as we move to higher confidence levels: around 3.1% at 90%, 4.4% at 95%, and about 6.0% at 99%.
That’s exactly what we expect higher confidence means we look further into the tail, so losses are rarer but more severe.</span></p>



<span style="color:#0b7a3e"><strong>methodological note:</strong> 
We recall that EVT-based Value at Risk relies on extrapolation beyond the observed sample using only the tail behavior of the distribution. As a result, the estimates become more sensitive as the confidence level increases</span>


<span style="color:#7a3e0b">Compared to the historical VaR computed in Question A the EVT-based VaR has greater weight on the tail of the loss distribution. As a result, it leads to higher risk estimates at very high confidence levels.
The EVT-based VaR has a limited number of extreme observations and can be sensitive to the choice of the sequence k and to sampling variability.</span>

#### II. Bouchaud price impact model: definition and parameters

We now introduce the price impact model proposed by Bouchaud, which will be used to estimate the impact of trades on prices.

The main idea is that price changes can be written as the cumulative effect of past trades where each trade contributes to the price movement (signed impact that decays over time). This allows us to capture both the direction of trades and the persistence of their impact.

<span style="color:#1f5faa"><strong>Course reference:</strong>In this question, we use the transitory impact model presented in the course, where prices are described as the cumulative effect of past signed trades (see slide page 133 of the lecture notes).</span>  

Parameter interpretation: 

The transitory impact model relies on a small number of parameters that jointly determine how trades affect prices both in magnitude and over time.  

- The parameter V is the overall scale of price impact: higher values of V imply that individual trades tend to move prices more strongly.

- The exponent r controls how impact depends on traded volume. When r is smaller than 1, the impact function is concave(commonly observed in empirical studies)

- gamma determines how persistent the impact of past trades is. A larger gamma means a faster decay of impact, while smaller values imply a longer memory in prices.

Finally, the parameter lambda introduced in the TD4 framework is used to control how past information is weighted over time.

#### III. Estimation strategy

We want to recover the parameters that both have an impact on the magnitude and the persistence of price by analising the relationship between price changes and past signed trades. At this stage, we focus on the general estimation logic before turning to the practical implementation.

The underlying idea is that short term price variations can be explained by the cumulative effect of past trades, as predicted by the model. In practice, observed price changes are related to past values through a parametric specification of the impact decay.

The parameters of the model are then estimated by minimizing the deviation between observed price changes and those implied by the model.

<span style="color:#1f5faa"><strong>Course reference:</strong> This estimation approach follows the methodology introduced in TD4, which relies on a parametric specification of the impact kernel and a regression framework.</span>  

##### Implementation details
We first construct the price increments that will be used as the dependent variable in the estimation.

```python
df_est = df_vars.copy()

df_est["delta_price"] = df_est["price"].diff() #we compute price increments from the transaction price

df_est[["time", "price", "delta_price"]].head()
```

##### Specification of the impact kernel

In order to estimate the model, we need to specify a parametric form for the impact decay function.  
Following the TD4 framework, we assume a power-law decay of impact over time.

<span style="color:#1f5faa"><strong>Course reference:</strong> In TD4, the impact kernel is sdescribed as a decreasing power law function of the lag( captures the gradual decay of price impact over time)</span>  

```python
max_lag = 50
lags = np.arange(1, max_lag + 1) #  lag structure used for the impact kernel

#parametric pow law form of the impact kernel
def impact_kernel(lags, gamma):
    return lags ** (-gamma)
```

##### Parameter estimation

For a given pair (r, gamma), we construct a predictor based on past signed trades and the impact kernel and we estimate the scaling parameter V using ordinary least squares.  
We then explore a simple grid of values for r and gamma and we keep the combination that provides the best fit (mean squared error)

```python
#we only keep trades with known volume here.
df_fit = df_est.copy()
df_fit = df_fit.dropna(subset=["delta_price", "eps", "spread", "volume"]).reset_index(drop=True)

df_fit = df_fit[df_fit["eps"].isin([-1, 1])]
df_fit = df_fit[(df_fit["spread"] >= 0) & (df_fit["volume"] >= 0)]

max_lag = 50 # We keep a reasonable maximum lag for the kernel 
lags = np.arange(1, max_lag + 1)

def impact_kernel_power(lags, gamma):
    # impact decay: G(l) = l^{-gamma}
    return lags ** (-gamma)

def build_regressor(df, r, gamma, max_lag=50):
    """
    We build X_t and we model: delta_price_t ≈ V * X_t
    """
    G = impact_kernel_power(np.arange(1, max_lag + 1), gamma)

    signed_term = df["eps"] * df["spread"] * (df["volume"] ** r)
    X = np.zeros(len(df), dtype=float)
    for l in range(1, max_lag + 1):
        X += G[l - 1] * signed_term.shift(l).fillna(0.0).to_numpy()

    Y = df["delta_price"].to_numpy()
    return X, Y

def estimate_V_ols(X, Y):
    #  OLS 
    denom = np.dot(X, X)
    if denom == 0:
        return np.nan
    return np.dot(X, Y) / denom

def mse(Y, Y_hat):
    return np.mean((Y - Y_hat) ** 2)
# grid search
r_grid = np.arange(0.0, 1.01, 0.1) # concavity exponent
gamma_grid = np.arange(0.1, 2.01, 0.1) # decay exponent

best = {"mse": np.inf, "V": None, "r": None, "gamma": None} # structure using the best result

for r in r_grid:
    for gamma in gamma_grid:
        X, Y = build_regressor(df_fit, r=r, gamma=gamma, max_lag=max_lag)
        valid = np.isfinite(X) & np.isfinite(Y)
        Xv, Yv = X[valid], Y[valid] # build regression variables

        V_hat = estimate_V_ols(Xv, Yv) # estimate parameters
        if np.isnan(V_hat):
            continue
        Y_hat = V_hat * Xv
        current_mse = mse(Yv, Y_hat)

        if current_mse < best["mse"]: # jeep the best parameters
            best = {"mse": current_mse, "V": V_hat, "r": r, "gamma": gamma}

best
```

```python
pd.DataFrame([best])
```

<span style="color:#0b7a3e">The estimated  r is below one meaning a concave dependence of price impact on traded volume. The estimated decay exponent γ is very high implying that trade impact fades quickly and is mostly temporary.

This outcome may reflect a liquid market environment but it may also highlight limitations of the parametric specification or of the chosen lag window.</span>

<strong>Conclusion:</strong>  Using a simple parametric specification of Bouchaud’s transitory impact model we obtain estimates for the concavity and decay of price impact. The results confirm a concave dependence on traded volume and a transitory impact of trades .  
While the model captures key  features of market impact, the boundary estimate for the decay parameter highlights potential limitations of the specification. 

<span style="color:#7a3e0b"> Note: Gamma is at the top of our grid (gamma = 2.0), so the decay might be even faster, this looks like a boundary solution.</span>

#### II - Multiresolution correlations using Haar wavelets (Question E.a)

In this part, we look at how the three FX rates move together depending on the time scale. Using Haar wavelets, we compute multiresolution correlations between GBPEUR,SEKEUR and CADEUR, and we will discuss if an Epps effect appear.The idea is to separate very short term moves from long term moves. Then we compute correlations scale by scale (level by level) instead of using only one single correlation.

<span style="color:#1f5faa"><strong>Course reference:</strong> Following the lecture notes (page 291), we use the multiresolution framework based on the Haar wavelet transform to study how correlations between FX returns change across time scales.</span>

We then take the three return series and drop missing values and apply a Haar multireso lution decomposition. At each level, the function averages pairs of points and extracts local fluctuations. Its then repeats this procedure on the average to describe the signal at
increasingly longer time horizons. For each level, we compute correlations between the three FX pairs using the detail
coefficients. To compare those levels, we standardize each level series(mean of 0 ,standard dev of 1) before computing correlations

```python
#beefore we computing correlations, we quickly check that the three FX series are aligned
# on the same timestamps (row by rows). If not we would need to merge on a common time index.
aligned_gbp_sek = (df_clean["Date_GBPEUR"].values == df_clean["Date_SEKEUR"].values).all()
aligned_gbp_cad = (df_clean["Date_GBPEUR"].values == df_clean["Date_CADEUR"].values).all()

print("GBPEUR vs SEKEUR aligned:", aligned_gbp_sek)
print("GBPEUR vs CADEUR aligned:", aligned_gbp_cad)
```

```python
# we take the clean 15min returns and drop NaN so the three series stay perfect aligned
returns = df_clean[["Ret_GBPEUR", "Ret_SEKEUR", "Ret_CADEUR"]].dropna(how="any")

def haar_wavelet_transform(x, levels=3):

    details_per_level = []
    current = x.copy()

    for _ in range(levels):
        current = current[: (len(current) // 2) * 2] # Haar keep only an even number of points for Haar pairs
        approx = (current[::2] + current[1::2]) / 2
        detail = (current[::2] - current[1::2]) / 2 # Haar averaging and differencing
        details_per_level.append(detail)
        current = approx  # and we keep going on the smoother part

    return details_per_level

wavelet_results = {} # we compute the Haar coeffs for each FX return series
for col in returns.columns:
    wavelet_results[col] = haar_wavelet_transform(returns[col].values, levels=3)

correlations = {} #  we compute correlations between the detail coefficients
n_levels = len(wavelet_results["Ret_GBPEUR"])

for level in range(n_levels):
    gbp = wavelet_results["Ret_GBPEUR"][level]
    sek = wavelet_results["Ret_SEKEUR"][level]
    cad = wavelet_results["Ret_CADEUR"][level]
# we standardize (mean 0 and standard dev 1) so we can compare levels fairly
    gbp = (gbp - gbp.mean()) / gbp.std()
    sek = (sek - sek.mean()) / sek.std()
    cad = (cad - cad.mean()) / cad.std()

    label = f"Level {level + 1}"
    correlations[label] = {
        "GBPEUR-SEKEUR": float(np.corrcoef(gbp, sek)[0, 1]),
        "GBPEUR-CADEUR": float(np.corrcoef(gbp, cad)[0, 1]),
        "SEKEUR-CADEUR": float(np.corrcoef(sek, cad)[0, 1]),
    }

for lvl, corr in correlations.items():
    print(f"{lvl} correlations:") #clean print
    for pair, val in corr.items():
        print(f"  {pair}: {val:.4f}")
```

<span style="color:#1f5faa">Following the lecture notes (page 291), we apply a Haar multiresolution decomposition to the FX returns. The idea is to study correlations level by level, in order to see whether dependence looks similar at very short horizons or becomes clearer as we move to longer time scales.</span>

<span style="color:#7a3e0b">With 15-minute data, level 
𝑗
j roughly corresponds to a time horizon of about (2j)×15 minutes (so Level 1 ≈ 30 minutes, Level 2 ≈ 1 hour, and Level 3 ≈ 2 hours). The Haar “detail” coefficients capture price movements occurring at that specific scale, so they can be viewed as a simple scale-specific filter rather than a mere aggregation of returns.</span>

<span style="color:#2e7d32">At Level 1, correlations are fairly small overall: around 0.16 for GBPEUR–SEKEUR, 0.17 for SEKEUR–CADEUR, and 0.22 for GBPEUR–CADEUR. At this shortest horizon, the three series therefore do not move together very strongly, which is typical for high-frequency data where noise and small timing effects tend to obscure dependence.</span>

<span style="color:#2e7d32">At Level 2, dependence becomes more visible for some pairs. The correlation between GBPEUR and CADEUR increases to about 0.33, while SEKEUR–CADEUR rises to roughly 0.25. In contrast, GBPEUR–SEKEUR remains lower at around 0.14. This suggests that CADEUR shares more common movements with the other exchange rates at this intermediate horizon.</span>

<span style="color:#2e7d32">At Level 3, the ranking changes again. GBPEUR–SEKEUR becomes the most correlated pair at about 0.30, while GBPEUR–CADEUR decreases to around 0.23 and SEKEUR–CADEUR falls to about 0.16. The pattern is therefore not perfectly monotone across levels, which is not surprising since each Haar level focuses on a different time scale rather than a simple cumulative aggregation.</span>

<span style="color:#7a3e0b">One natural interpretation is provided by the well-known Epps effect. At very short horizons, correlations often appear weaker, and they tend to emerge more clearly once we move to slightly longer time scales.</span>

<span style="color:#7a3e0b">Overall, correlations look weakest at the highest frequency and become easier to detect as we move to longer horizons, which is consistent with the presence of market microstructure effects at short time scales.</span>

<span style="color:#7a3e0b">Conclusion: In our sample, dependence mainly becomes visible once we move away from the finest scale. We also observe that the most correlated pair can change across levels, confirming that correlation is strongly scale dependent.</span>

We now turn to the estimation of the Hurst exponent to study the scaling properties of each FX rate.

#### III - Estimation of Hurst exponent and volatility scaling (Question E.b)


In this second part, we focus on the scaling properties of FX returns. We estimate the Hurst exponent for GBPEUR, SEKEUR and CADEUR in order to assess whether the return series exhibit persistence, anti-persistence, or behave close to a random walk. Based on the estimated Hurst exponents, we then compute the corresponding annualized volatilities.

<span style="color:#1f5faa"><strong>Course reference:</strong> We briefly recall the theoretical framework underlying the estimation of the Hurst exponent, as presented in the lecture notes (page 251 - 252).</span>

<span style="color:#1f5faa">According to the course we have that :</span>

We now implement this estimator to compute the Hurst exponent for each FX rate.

```python
def Hurst_Exponent(X):
    k = 2  # we use the moment of order 2 (same estimator as in the slides)
    T = len(X)

    # quick safety check (we need enough points)
    if T < 4:
        return np.nan

    M_2 = 0     # we compute M_2 with step size = 1
    for i in range(1, T):
        M_2 += abs(X[i] - X[i - 1])**k
    M_2 /= (T - 1)

    # we compute M_2_prime with step size = 2
    M_2_prime = 0
    for i in range(1, T // 2):
        M_2_prime += abs(X[2 * i] - X[2 * i - 2])**k
    M_2_prime /= (T // 2 - 1)

    #  apply H = 0.5 * log2(M_2_prime / M_2)
    return 0.5 * np.log2(M_2_prime / M_2)


# we take the 15 min returns from our cleaned dataset
ret_gbp = returns["Ret_GBPEUR"].values.astype(float)
ret_sek = returns["Ret_SEKEUR"].values.astype(float)
ret_cad = returns["Ret_CADEUR"].values.astype(float)

# we build X_t by cumulating returns (price)
X_GBPEUR = np.cumsum(ret_gbp)
X_SEKEUR = np.cumsum(ret_sek)
X_CADEUR = np.cumsum(ret_cad)

# we compute H for each FX rate
Hurst_GBPEUR = Hurst_Exponent(X_GBPEUR)
Hurst_SEKEUR = Hurst_Exponent(X_SEKEUR)
Hurst_CADEUR = Hurst_Exponent(X_CADEUR)

print("Hurst exponent for GBPEUR:", Hurst_GBPEUR)
print("Hurst exponent for SEKEUR:", Hurst_SEKEUR)
print("Hurst exponent for CADEUR:", Hurst_CADEUR)
```

<span style="color:#1f5faa"><strong>Course reference:</strong> Hurst exponent and volatility scaling(page 210): The Hurst exponent is a measure of the scaling behavior of a time series and provides
information about deviations from a pure Brownian motion. A value of \( H = 0.5 \)
corresponds to a standard Brownian motion, while values different from 0.5 indicate
persistence or anti-persistence.</span>

<span style="color:#2e7d32">Based on the estimated Hurst exponents, which are larger than 0.5, the cumulated return process X_t (built from intraday returns) shows signs of persistence compared to a Brownian motion (H = 0.5). We still interpret these values with caution because the estimator relies on scaling assumptions and our estimate is based on intraday data with a finite sample and possible microstructure effects.
</span>

<span style="color:#1f5faa"><strong>Course reference:</strong> Instead of the usual square-root-of-time rule (which assumes independent returns), we follow the course scaling law σ(τ) ∝ τ^H (slides, p. 210). With 15-minute returns, this gives the annualized volatility below.</span>

<span style="color:#1f5faa"><strong>Scaling law (Hurst):</strong> σ(Δt) ∝ (Δt)^H ⇒ σ(T) = σ(Δt) * (T/Δt)^H.</span>

<span style="color:#1f5faa"><strong>Annualisation (Δt = 15 min):</strong> σ_annual = σ_15min * N^H, with N = 252 * 10 * 4 (252 days/year, 10 hours/day, 4*15min/hour).</span>

<span style="color:#1f5faa"><strong> Final :</strong> σ_annual = σ_15min * (252 * 10 * 4)^H</span>

```python
def calculate_annualized_volatility(returns_series, hurst_exponent):
    # first we just take the average return on our sample
    mean_return = sum(returns_series) / len(returns_series)

    # then we compute the volatility at the 15 min step 
    std_return = np.sqrt(
        sum((x - mean_return) ** 2 for x in returns_series) / (len(returns_series)))

# now we scale it up to one year using the Hurst exponent (so we don't rely on independence)
    return std_return * (252 * 10 * 4) ** hurst_exponent# for our dataset, we take 252 trading days per year and about 10 trading hours per day(4 point per hour)


annualy_vol_GBPEUR = calculate_annualized_volatility(
    returns["Ret_GBPEUR"].dropna().values.astype(float),
    Hurst_GBPEUR) #we apply the same formule to each FX returns series
print(f"Annualized volatility for GBPEUR: {annualy_vol_GBPEUR:.6f}")

annualy_vol_SEKEUR = calculate_annualized_volatility(
    returns["Ret_SEKEUR"].dropna().values.astype(float),
    Hurst_SEKEUR)
print(f"Annualized volatility for SEKEUR: {annualy_vol_SEKEUR:.6f}")

annualy_vol_CADEUR = calculate_annualized_volatility(
    returns["Ret_CADEUR"].dropna().values.astype(float),
    Hurst_CADEUR)
print(f"Annualized volatility for CADEUR: {annualy_vol_CADEUR:.6f}")
```

Conclusion (E.b): We find similar Hurst exponents for the three FX return series with values above 0.5. This suggests some degree of persistence meaning the results should be interpreted with caution due to intraday data, finite sample effects and the estimation method. In practice, we estimate H on the cumulative sum of intraday returns(gives a priceprocess). Indeed, this is a convenient proxy even if it may differ from estimating H directly on log-prices. Using H scaling then allows us to annualize volatility beyond the standard square-root-of-time rule.

Conclusion: Overall, our results show that FX behavior depends on the time scale considered --> should be taken into account when analyzing risk.
